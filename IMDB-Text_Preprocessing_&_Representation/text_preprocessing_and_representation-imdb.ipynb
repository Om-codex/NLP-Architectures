## **Removing HTML Tags**

```python
import re
def remove_html_tags(text):
    if not isinstance(text, str):
        return text
    return re.sub(r'<[^>]+>', '', text)
```

```python
text = "<html><body><p>Movie 1</p> <p>Actor - Aamir Khan</p><p>Click here to watch <a href='https://google.com/'></body></html>"
```

```python
remove_html_tags(text)
```

```python
df['overview'].apply(remove_html_tags)
```

## **Removing URL's**

```python
def remove_url(text):
    if not isinstance(text, str):
        return text
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
df['overview']=df['overview'].apply(remove_url)
```

## **Removing Punctuation**

```python
import string,time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punctuation(text):  # slow function
    for char in exclude:
        text = text.replace(char,'')
    return text
```

```python
text = "string. with.74 Punctuation?"
start = time.time()
remove_punctuation(text)
time1 = time.time() - start
print(time1)
```

```python
def remove_punc1(text):   # fast function
    if not isinstance(text, str):
        return text
    return text.translate(str.maketrans('','',exclude))
```

```python
# start = time.time()
remove_punc1(text)
time2 = time.time() - start
# print(time1)
```

```python
time1/time2
```

```python
df['overview']= df['overview'].apply(remove_punc1)
df['overview']
```

## **Removing Stopwords**

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
STOP_WORDS = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    if not isinstance(text,str):
        return ""
    
    return " ".join(
        word for word in text.split() if word.lower() not in STOP_WORDS
    )
```

```python
remove_stopwords('Hello everyone this is Om Codex a Data Science and ML learning enthusiast while learning I try to code and implement the logic as well and everytime I learn new concept I always try to build a end-to-end project from scratch by understanding each of the steps.')
```

```python
df.head()
```

```python
df['overview']= df['overview'].apply(remove_stopwords)
df['overview']
```

## **Removing Emoji If any**

```python
import re
def remove_emoji(text):
    if not isinstance(text,str):
        return ""
    
    emoji_pattern = re.compile("[" u"\U0001F600-\U0001F64F" # EMOTIONS
                                   u"\U0001F300-\U0001F5FF" # SYMBOLS & PICTOGRAPHS
                                   u"\U0001F680-\U0001F6FF" # TRANSPORT & MAP SYMBOLS
                                   u"\U0001F1E0-\U0001F1FF" #FLAGS
                                   u"\U00002702-\U000027B0"
                                   u"\U000024C2-\U0001F251"
                                   "]+",flags = re.UNICODE)
    return emoji_pattern.sub(r'',text)
```

```python
remove_emoji("Loved the movie 😘😘")
```

```python
df['overview'] = df['overview'].apply(remove_emoji)
df['overview']
```

## **Stemming**

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

## **Lemmatization**

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Initialize lemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

# Input sentence
sentence = (
    "He was running and eating at same time. "
    "He has bad habit of swimming after playing long hours in the Sun."
)

# Define punctuation
punctuations = "?.,!"

# Tokenize
sentence_words = word_tokenize(sentence)

# Remove punctuation safely
sentence_words = [word for word in sentence_words if word not in punctuations]

# Print header
print("{:20} {:20}".format("Word", "Lemma"))
print("-" * 40)

# Lemmatize and print
for word in sentence_words:
    print("{:20} {:20}".format(word, wordnet_lemmatizer.lemmatize(word,pos = 'v')))
```
