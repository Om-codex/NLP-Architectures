```python
import re 
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
```

```python
def cleanup(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "",text)
    tokens = word_tokenize(text)

    return tokens
```

```python
df['q1_tokens'] = df['question1'].apply(cleanup)
df['q2_tokens'] = df['question2'].apply(cleanup)
```

```python
df['q1_len'] = df['q1_tokens'].apply(len)
df['q2_len'] = df['q2_tokens'].apply(len)
df.head()
```

```python
from collections import Counter

all_tokens = []

for tokens in df['q1_tokens']:
    all_tokens.extend(tokens)
for tokens in df['q2_tokens']:
    all_tokens.extend(tokens)

token_counts = Counter(all_tokens)

MAX_VOCAB = 20000

most_common = token_counts.most_common(MAX_VOCAB - 2)

vocab_words = [w for w, _ in most_common]

word2idx = {"<PAD>": 0, "<UNK>":1}
for i, w in enumerate(vocab_words, start = 2):
    word2idx[w] = i

idx2word = {i: w for w, i in word2idx.items()}

vocab_size = len(word2idx)
print("Vocab Size: ",vocab_size)
```

```python
def tokens_to_ids(tokens, word2idx):
    return [word2idx.get(t, word2idx["<UNK>"]) for t in tokens]

df["q1_ids"] = df["q1_tokens"].apply(lambda x: tokens_to_ids(x, word2idx))
df["q2_ids"] = df["q2_tokens"].apply(lambda x: tokens_to_ids(x, word2idx))
```

```python
import numpy as np

MAX_LEN = 20

def pad_truncate(seq, max_len=MAX_LEN, pad_value=0):
    if len(seq) >= max_len:
        return seq[:max_len]
    else:
        return seq + [pad_value] * (max_len - len(seq))

df["q1_seq"] = df["q1_ids"].apply(pad_truncate)
df["q2_seq"] = df["q2_ids"].apply(pad_truncate)
```

```python
X1 = np.array(df["q1_seq"].tolist(), dtype=np.int64)
X2 = np.array(df["q2_seq"].tolist(), dtype=np.int64)
y  = df["is_duplicate"].values.astype(np.float32)

print(X1.shape, X2.shape, y.shape)
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
```

```python
class QuoraDataset(Dataset):
    
    def __init__(self, X1, X2, y):
        self.X1 = torch.tensor(X1, dtype=torch.long)
        self.X2 = torch.tensor(X2, dtype=torch.long)
        self.y  = torch.tensor(y, dtype=torch.float32)
        
    def __len__(self):
        return len(self.y)
    
    def __getitem__(self, idx):
        return self.X1[idx], self.X2[idx], self.y[idx]
```

```python
from sklearn.model_selection import train_test_split

X1_train, X1_val, X2_train, X2_val, y_train, y_val = train_test_split(
    X1, X2, y, test_size=0.2, random_state=42
)
```

```python
train_dataset = QuoraDataset(X1_train, X2_train, y_train)
val_dataset   = QuoraDataset(X1_val, X2_val, y_val)
```

```python
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=64)
```

```python
import torch.nn as nn
```

```python
class SiameseLSTM(nn.Module):
    
    def __init__(self, vocab_size, embed_dim, hidden_dim):
        super().__init__()
        
        # Embedding Layer
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        
        # Shared LSTM Encoder
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        
        # Final Classifier
        self.fc = nn.Linear(hidden_dim, 1)
        
    def encode(self, x):
        
        # x → embeddings
        emb = self.embedding(x)
        
        # emb → LSTM
        _, (hidden, _) = self.lstm(emb)
        
        # last hidden state
        return hidden[-1]
    
    def forward(self, q1, q2):
        
        v1 = self.encode(q1)
        v2 = self.encode(q2)
        
        diff = torch.abs(v1 - v2)
        
        output = self.fc(diff)
        
        return torch.sigmoid(output).squeeze()
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = SiameseLSTM(
    vocab_size=vocab_size,
    embed_dim=100,
    hidden_dim=128
).to(device)
```

```python
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
def train_epoch(model, loader):
    
    model.train()
    total_loss = 0
    
    for q1, q2, labels in loader:
        
        q1 = q1.to(device)
        q2 = q2.to(device)
        labels = labels.to(device)
        
        preds = model(q1, q2)
        
        loss = criterion(preds, labels)
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
    
    return total_loss / len(loader)
```

```python
def eval_epoch(model, loader):
    
    model.eval()
    total_loss = 0
    
    with torch.no_grad():
        for q1, q2, labels in loader:
            
            q1 = q1.to(device)
            q2 = q2.to(device)
            labels = labels.to(device)
            
            preds = model(q1, q2)
            loss = criterion(preds, labels)
            
            total_loss += loss.item()
    
    return total_loss / len(loader)
```

```python
epochs = 5

for epoch in range(epochs):
    
    train_loss = train_epoch(model, train_loader)
    val_loss = eval_epoch(model, val_loader)
    
    print(f"Epoch {epoch+1}")
    print("Train Loss:", train_loss)
    print("Val Loss:", val_loss)
```

```python
model.eval()

for q1, q2, labels in val_loader:
    
    q1 = q1.to(device)
    q2 = q2.to(device)
    
    preds = model(q1, q2)
    
    preds = (preds > 0.5).cpu().numpy()
    
    for i in range(len(preds)):
        if preds[i] != labels[i]:
            print("Q1:", df.iloc[i]["question1"])
            print("Q2:", df.iloc[i]["question2"])
            print("Actual:", labels[i].item())
            print("Pred:", preds[i])
            print()
    
    break
```
